```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### 1. 创建表格

```python
workout_dict = {
    "calories":[420,380,390,390],
    "duration":[50,40,45,45],
    "type":['run','walk','walk','run']
}

workout = pd.DataFrame(workout_dict)
display(workout)
```

### 2. DataFrame 单独指定/改变索引

```python
# 直接重新赋值行索引
workout.index=["day1","day2","day3","day4"]
display(workout)
```

```python
# 直接重新赋值列索引
workout.columns=["calories","duration","type"]
workout
```

```python
# 在构建Pandas时单独指定索引
workout = pd.DataFrame(workout_dict,index=["day1","day2","day3","day4"])
display(workout)
```

### 3. 访问表格中的列与行

#### --- 选择索引与指定列 ---

```python
# 选择一列 []
print(workout['calories'])
# 选择一列（保留列表格式） [[]]
workout[['calories']]
```

#### --- 根据索引选取数据 `loc`---

```python
# loc:
workout.loc['day1',:]
```

```python
# 保留 DataFrame 格式
workout.loc[['day1'],:]
```
